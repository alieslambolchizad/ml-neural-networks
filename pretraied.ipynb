```python
from keras.applications import ResNet50
from keras.applications import VGG16
from keras.preprocessing.image import img_to_array
from keras.preprocessing.image import load_img
from keras.applications import imagenet_utils
import numpy as np
import matplotlib.pyplot as plt
```

```python
# model=ResNet50(weights='imagenet')
model=VGG16(weights='imagenet')
```

```python
img=load_img('Vulpes_vulpes_ssp_fulvus.jpg')
```

```python
type(img)
```

```python
img2 = img.resize((224, 224))
```

```python
img2
```

```python
im=img_to_array(img2)
```

```python

im.shape
```

```python
im2=imagenet_utils.preprocess_input(im.reshape(1,224,224,3))
plt.imshow(im2.reshape(224,224,3).astype('uint8'))
```

```python
p=model.predict(im2.reshape(1,224,224,3))
```

```python
p.shape
#1000 tedad class haei ke dar imagenet bod
#1 tedad nemone haye ke behesh dadim ta preidict konad
```

```python
imagenet_utils.decode_predictions(p)
#avalin item shomare class to dataset imagenet
# dovomi label
#sevomi ehtemal bodan dar label
```

```python
res=imagenet_utils.decode_predictions(p)[0][0]
```

```python
print(f'Result: {res[1]} ({res[2]*100:.2f}%)')
```
