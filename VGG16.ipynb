```python
model=VGG16(weights='imagenet',include_top=False,input_shape=(48,144,3),pooling='none')
```

```python
for l in model.layers:
  l.trainable=False   #yani in laye toye farayand learning sherkat nemikonad
```

```python
a = Flatten()(model.output)
a = Dense(64, activation='relu')(a)
a = Dense(32, activation='softmax')(a)

model2 = Model(inputs=model.input, outputs=a)
```

```python
model2.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy']
              )
```

```python
import os
from glob import glob
def datagenerator(path,batch_size=100,c=0):
  files=glob(os.path.join(path,'*.png'))
  offset=0
  while offset<len(files):
    items=files[offset:offset+batch_size]
    offset+=batch_size
    x=[]
    y=[]
    for f in  items:
      n=os.path.basename(f)
      n=os.path.splitext(n)[0]
      n=list(n)[c]
      im=cv2.imread(f)
      im=cv2.cvtColor(im,cv2.COLOR_BGR2RGB)
      im=cv2.resize(im,(144,48))
      x.append(im)
      y.append(to_categorical(encoder.transform([n])
      ,num_classes=len(encoder.classes_)))
    x=np.array(x,ndmin=4)
    y = np.vstack(y)

    yield x,y
```
